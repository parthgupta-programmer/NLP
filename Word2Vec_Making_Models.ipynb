# Making own Word2Vec models

```python
import gensim
from gensim.models import Word2Vec,KeyedVectors
```

```python
sentences = [
    ["machine", "learning", "is", "a", "field", "of", "artificial", "intelligence"],
    ["artificial", "intelligence", "enables", "machines", "to", "learn"],
    ["machine", "learning", "models", "learn", "from", "data"],
    ["deep", "learning", "is", "a", "subset", "of", "machine", "learning"],
    ["deep", "learning", "uses", "artificial", "neural", "networks"],
    ["neural", "networks", "are", "used", "in", "deep", "learning"],
    ["supervised", "learning", "uses", "labeled", "training", "data"],
    ["unsupervised", "learning", "finds", "patterns", "in", "data"],
    ["reinforcement", "learning", "uses", "rewards", "and", "penalties"],

    ["python", "is", "a", "programming", "language"],
    ["python", "is", "widely", "used", "in", "machine", "learning"],
    ["python", "libraries", "help", "developers", "build", "applications"],
    ["numpy", "is", "used", "for", "numerical", "computing"],
    ["pandas", "is", "used", "for", "data", "analysis"],
    ["matplotlib", "helps", "visualize", "data"],
    ["scikit", "learn", "provides", "machine", "learning", "algorithms"],
    ["tensorflow", "is", "used", "for", "deep", "learning"],
    ["pytorch", "is", "a", "deep", "learning", "framework"],

    ["data", "science", "combines", "statistics", "programming", "and", "domain", "knowledge"],
    ["data", "scientists", "analyze", "large", "datasets"],
    ["data", "analysis", "helps", "discover", "useful", "insights"],
    ["data", "visualization", "makes", "complex", "information", "understandable"],
    ["statistics", "helps", "analyze", "data"],
    ["big", "data", "requires", "efficient", "processing", "techniques"],

    ["natural", "language", "processing", "is", "a", "field", "of", "artificial", "intelligence"],
    ["natural", "language", "processing", "helps", "computers", "understand", "human", "language"],
    ["word", "embeddings", "represent", "words", "as", "vectors"],
    ["word2vec", "learns", "word", "embeddings", "from", "context"],
    ["tokenization", "splits", "text", "into", "smaller", "units"],
    ["sentiment", "analysis", "identifies", "emotions", "in", "text"],
    ["transformers", "are", "used", "in", "modern", "language", "models"],

    ["students", "learn", "programming", "through", "practical", "projects"],
    ["education", "helps", "students", "develop", "technical", "skills"],
    ["programming", "requires", "logical", "thinking", "and", "problem", "solving"],
    ["developers", "write", "code", "to", "solve", "problems"],
    ["software", "engineering", "involves", "designing", "and", "building", "software"],
    ["algorithms", "are", "used", "to", "solve", "computational", "problems"]
]

# Create the Word2Vec model
model=Word2Vec(sentences, window=5, min_count=1, vector_size=100)

# Build vocabulary
model.build_vocab(sentences)

# Train the model
model.train(sentences,total_examples=model.corpus_count,epochs=model.epochs)
```

```python
model.wv.most_similar("python")
```

```python
model.wv.most_similar("learning")
```

```python
print(model.wv.similarity("machine", "deep"))
print(model.wv["intelligence"])
```
