## 2.Text Classification
### -> It is used to categorize text into groups based on labeled data.
### -> We can input vectorized text data into classification algorithm(Naive Bayes)

#### GOAL: Predict which reviews are high priority (vs low priority) that we need to address right away

```python
import pandas as pd

df2=pd.read_excel(r'data/Popchip_Reviews.xlsx')
df2.head()
```

```python
df2.shape
```

```python
df2['Priority'].value_counts()
```

```python
# Text Preprocessing on Text column

from defined_nlp_functions import clean_and_normalize

df2['Cleaned_Data']=clean_and_normalize(df2['Text']) # Using clean_and_normalize function
df2[['Text','Cleaned_Data']]
```

### Naive Bayes with Count Vectorizer


```python
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer(stop_words='english',ngram_range=(1,2),min_df=.15)
X=cv.fit_transform(df2['Cleaned_Data'])
X_df=pd.DataFrame(X.toarray(),columns=cv.get_feature_names_out())
X_df
```

```python
y=df2[['Priority']]
y.head()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X_df,y,test_size=0.3,random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB

nb_model=MultinomialNB()
nb_model.fit(X_train,y_train)
```

```python
nb_pred_train=nb_model.predict(X_train)
nb_pred_test=nb_model.predict(X_test)
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score,accuracy_score,classification_report

print('On Training Data:\n')
print(f'Classification Report:\n\n{classification_report(y_train,nb_pred_train)}')
print(f'Accuracy:{accuracy_score(y_train,nb_pred_train)}\n\n')

print(' * '*30)

print('\n\nOn Testing Data:\n')
print(f'Classification Report:\n\n{classification_report(y_test,nb_pred_test)}')
print(f'Accuracy:{accuracy_score(y_test,nb_pred_test)}')

```

#### Testing on New Reviews

```python
new_reviews=pd.DataFrame(['I love the snacks.','I really enjoyed using this product.','I didn\'t like the product.',"Pop chips are my favorite! I love these chips so much.",
    "Taste bad. I don't like the flavor options or taste.",
    "Solid snack."],columns=['New Review'])
new_reviews
```

```python
# Data Cleaning and Vectorizing the above reviews

new_reviews['Cleaned Review']=clean_and_normalize(new_reviews['New Review'])
new_reviews_df=pd.DataFrame((cv.transform(new_reviews['Cleaned Review'])).toarray(),columns=cv.get_feature_names_out())
new_reviews['Priority_NB']=nb_model.predict(new_reviews_df)
new_reviews
```
