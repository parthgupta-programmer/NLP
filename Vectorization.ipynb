## Count Vectorizer(Word Counts)

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_colwidth',None)
data = [
    "The product is AMAZING!! 😍 I absolutely love it.",
    "I ordered 2 bottles for $15 from Amazon.",
    "This phone case looks good, but the quality is terrible...",
    "WOW!!! Best purchase I've made this year 🔥🔥",
    "The delivery took 5 days, which was okay.",
    "I don't like this product — it's too expensive!",
    "Battery life is great, but the camera isn't that good.",
    "I've been using this for 3 months & it's still working perfectly.[GREAT PRODUCT]",
    "Worst experience ever!!! 😡 The product arrived damaged.",
    "Can someone tell me if this is compatible with iPhone 15?",
    "The headphones are REALLY good!!! Sound quality = 10/10 🎧",
    "I bought it yesterday... let's see how long it lasts.",
    "Not bad, not great. It's just OK.",
    "The seller sent the wrong color (Black instead of Blue).",
    "My sister LOVES this product, but I personally don't.",
]

data2 = [
    "When life gives you lemons, make lemonade! 🙂",
    "She bought 2 lemons for $1 at Maven Market.",
    "A dozen lemons will make a gallon of lemonade. [AllRecipes]",
    "lemon, lemon, lemons, lemon, lemon, lemons",
    "He's running to the market to get a lemon — there's a great sale today.",
    "iced tea is my favorite",
    "I didn't like the taste of that lemonade at all.",
    "My lemons went bad before I could use them, unfortunately.",
]

df2=pd.DataFrame(data2,columns=['Text'])
df=pd.DataFrame(data,columns=['Customer_Review'])

df
```

```python
from defined_nlp_functions import nlp_pipeline

df['Customer_Review']=nlp_pipeline(df['Customer_Review'])
df['Customer_Review']
```

```python
from sklearn.feature_extraction.text import CountVectorizer


# ---> ngram_range : It decides how many consecutive words (tokens) should be considered together.
    
## ngram_range=(1, 1): Means unigrams — individual words:

# "I love machine learning"

# → I
# → love
# → machine
# → learning

## ngram_range=(1, 2): Means unigrams + bigrams:

# I
# love
# machine
# learning
# I love
# love machine
# machine learning

## ngram_range=(1, 3): Means unigrams + bigrams + trigrams


# ---> min_df means minimum document frequency:  It removes words that occur in very few documents.A word must present at least {value} documents in the vocabulary.

cv=CountVectorizer(stop_words='english',ngram_range=(1,2),min_df=1)


dtm=cv.fit_transform(df['Customer_Review'])


dtm_df=pd.DataFrame(dtm.toarray(),columns=cv.get_feature_names_out())
dtm_df
```

```python
nlp_pipeline(df2['Text'])
```

```python
cv2=CountVectorizer(stop_words='english',ngram_range=(1,2),min_df=2)

dtm2=cv2.fit_transform(df2['Text'])
dtm2_df=pd.DataFrame(dtm2.toarray(),columns=cv2.get_feature_names_out())

dtm2_df
```

## TF-IDF Vectorizer: It emphasizes important words by reducing the impact of common words.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf=TfidfVectorizer(stop_words='english',ngram_range=(1,2),min_df=2)
dtm3=tfidf.fit_transform(df2['Text'])

dtm3_df=pd.DataFrame(dtm3.toarray(),columns=tfidf.get_feature_names_out())
dtm3_df
```
