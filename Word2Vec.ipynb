# Using Pre-Trained Models

```python
import gensim

from gensim.models import Word2Vec,KeyedVectors

model=KeyedVectors.load_word2vec_format(r"C:\Users\HP\Downloads\GoogleNews-vectors-negative300.bin.gz",binary=True,limit=50000)
```

```python
model.most_similar('cricket',topn=10)
```
