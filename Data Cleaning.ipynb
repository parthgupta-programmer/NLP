# <center>Natural Language Processing (NLP)</center>

NLP is the process of making computers work with human language.

For example: "The phone has an amazing camera but the battery is terrible."

An <b>NLP</b> system might need to determine:

What are the <b>words</b>?<br>
What is the <b>sentiment</b>?<br>
Which <b>words</b> are important?<br>
Is this review <b>positive or negative</b>?<br>
What <b>product/features</b> are being discussed?<br>

The computer doesn't naturally understand the sentence like a human. We need to convert and process the text so algorithms can work with it.

## Cleaning Function

```python
# The cleaning of the data depends upon the task.
# Here, for example in sentiment analysis- numbers are important and so emojis are.

def clean_data(series):
    
    # Lowercasing
    output=series.str.lower()
    
    # Removing Punctuation
    output=output.str.translate(str.maketrans('','',string.punctuation))
    
    # Removing Spaces 
    output=output.apply(lambda x: re.sub(r'\s+',' ',x)).str.strip()
    
    # Converting emojis to text
    output=output.apply(lambda x: emoji.demojize(x)).str.replace(':','')
    

    return output
```
