### Vectorization: It is the process of converting text data into numeric data so that future data analysis and machine learning techniques can be applied.

<b>Techniques</b>:<br>
    
1. <b>Word Counts</b><br>
2. <b>TF-IDF</b><br>
3. <b>Embeddings</b>
