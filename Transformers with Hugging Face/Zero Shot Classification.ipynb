# Zero Shot Classification

It is used to quickly categorize text without labels.<br>
• The default LLM for zero-shot classification is BART (encoder-decoder)

```python
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings('ignore')

df=pd.read_excel(r'C:\Users\HP\Desktop\PG\NLP\NLP with ML\data\Popchip_Reviews.xlsx')
df.head()
```

```python
from transformers import pipeline

classifier=pipeline('zero-shot-classification',
                        model='facebook/bart-large-mnli',
                        device=-1
                       )
```

You have to provide the label options and the model returns scores that classify it into one of those labels (adding up to 1).

```python
text='When life gives you lemons, make lemonade! 😋'
classifier(text,['quote','food & drinks','technology'])
```

```python
df['Category']=df.Text[:10].apply(lambda x: classifier(x, ['order', 'taste & texture', 'good', 'flavor', 'health'])['labels'][0])
```
