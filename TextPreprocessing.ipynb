### Token_Lemma_NonStop Function

```python
def token_lemma_nonstop(text):
    
    doc=nlp(text)
    output=[token.lemma_ for token in doc if not token.is_stop]
    output = ' '.join(output)
    return output
```

```python
from defined_nlp_functions import clean_data

df['Customer_Review']=clean_data(df['Customer_Review'])
df['Customer_Review']=df['Customer_Review'].apply(lambda x: token_lemma_nonstop(x))
```

### POS Function

```python
def pos(text,pos_list=['NOUN','PROPN']):
    doc=nlp(text)
    output=[token.text for token in doc if token.pos_ in pos_list]
    output=' '.join(output)
    return output
```

```python
df['Customer_Review'].apply(pos)
```

## NLP Pipeline

```python
def nlp_pipeline(series):
    output=clean_data(series)
    output=output.apply(token_lemma_nonstop)
    output=output.apply(pos)
    return output
```
